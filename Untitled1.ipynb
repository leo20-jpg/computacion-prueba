#### **2-(a)** Apóyate en el programa que vimos en clase y escribe un programa que muestre el comportamiento del mapeo logístico mediante una gráfica.


```python


import numpy as np
import matplotlib.pyplot as plt

r_valores = np.linspace(1,4,301)

valores_r = []
valores_x = []

for r in r_valores:

    x = 0.5

    for i in range(1000):
        x = r*x*(1-x)

    for i in range(1000):
        x = r*x*(1-x)

        valores_r.append(r)
        valores_x.append(x)

plt.figure(figsize=(10,7))

plt.plot(valores_r, valores_x, 'k.', markersize=1)

plt.title("Diagrama de Feigenbaum")
plt.xlabel("r")
plt.ylabel("x")
plt.grid()
plt.show()

```

#### **(a) Escribe un programa para crear una imagen del conjunto de Mandelbrot** realizando la iteración para todos los valores de $c = x + iy$ en una cuadrícula de $N \times N$ que abarque la región donde $-2 \le x \le 2$ y $-2 \le y \le 2$. Haz una gráfica de densidad (*density plot*) en el que los puntos de la cuadrícula dentro del conjunto de Mandelbrot estén coloreados en negro y los de afuera estén coloreados en blanco.


```python


import numpy as np
import matplotlib.pyplot as plt

N = 500

x = np.linspace(-2,2,N)
y = np.linspace(-2,2,N)

mandelbrot = np.zeros([N,N])

for i in range(N):
    for j in range(N):

        c = x[j] + 1j*y[i]
        z = 0

        for k in range(100):
            z = z*z + c

            if abs(z) > 2:
                break

        if abs(z) <= 2:
            mandelbrot[i,j] = 1

plt.figure(figsize=(8,8))

plt.imshow(mandelbrot, origin="lower",
           extent=[-2,2,-2,2], cmap="gray_r")

plt.title("Conjunto de Mandelbrot")
plt.xlabel("Parte real")
plt.ylabel("Parte imaginaria")
plt.show()

```
