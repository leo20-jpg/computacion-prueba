### **1. La constante de Madelung** 
#### Consideremos un atomo de sodio en el origen, i.e.  i = j = k = 0, y calculemos la constante de Madelung. Si el espaciado de los atomos en la red es a, entonces la distancia desde elorigen al atomo en la posicíon (i, j, k) es: $$\sqrt{(ia)^2 + (ja)^2 + (ka)^2} = a\sqrt{i^2 + j^2 + k^2}$$
#### y el potencial en el origen creado por tal atomo es: $$V(i, j, k) = \pm \frac{e}{4\pi\epsilon_0 a \sqrt{i^2 + j^2 + k^2}}$$
#### siendo ϵ0 la permitividad del vac´ıo y el signo de la expresion se toma dependiendo de si i + j + k es par o impar. Así entonces, el potencial total que siente el atomo de sodio es la suma de esta cantidad sobre todos los demás atomos. Supongamos una caja cúbica alrededor del atomo de sodio en el origen, con L atomos en todas las direcciones, entonces: $$V_{\text{total}} = \sum_{\substack{i,j,k=-L \\ i,j,k \neq 0}}^{L} V(i, j, k) = \frac{e}{4\pi\epsilon_0 a} M$$
#### donde M es la constante de Madelung


####  Tecnicamente, la constante de Madelung es el valor de M cuando L → ∞, pero se puedeobtener una buena aproximacion simplemente usando un valor grande de L. **a)** Escribe un programa para calcular e imprimir la constante de Madelung para el cloruro de sodio. Utiliza un valor de L tan grande como puedas, sin dejar que tu programa se ejecutar en un tiempo razonable (un minuto o menos).

```python
from math import sqrt

L = 100

M = 0

for i in range(-L,L+1):
    for j in range(-L,L+1):
        for k in range(-L,L+1):

            if i == 0 and j == 0 and k == 0:
                continue

            r = sqrt(i**2 + j**2 + k**2)

            if (i+j+k)%2 == 0:
                M = M + 1/r
            else:
                M = M - 1/r

print("La constante de Madelung es aproximadamente:",M)
```

### **3. Coeficientes binomiales**

#### El coeficiente binomial $\binom{n}{k}$ es un número entero igual a: $$\binom{n}{k} = \frac{n!}{k!(n-k)!} = \frac{n \times (n-1) \times (n-2) \times \dots \times (n-k+1)}{1 \times 2 \times \dots \times k}$$
$\text{donde } k \ge 1, \quad \text{o bien } \binom{n}{0} = 1 \quad \text{cuando } k = 0$

#### a) Utiliza esta fórmula para escribir una función llamada `binomial(n,k)` (o como tú quieras) que calcule el coeficiente binomial para un $n$ y $k$ dados. Asegúrate de que tu función devuelva la respuesta en forma de un número entero (no flotante) y proporcione el valor correcto de 1 para el caso en que $k = 0$.

```python
def binomial(n,k):

    if k == 0:
        return 1

    numerador = 1
    denominador = 1

    for i in range(1,k+1):
        numerador = numerador*(n-i+1)
        denominador = denominador*i

    resultado = numerador//denominador

    return resultado


print(binomial(5,3))
```

#### b) Usando tu función, escribe un programa que imprima las primeras 20 líneas del "triángulo de Pascal". La $n$-ésima línea del triángulo de Pascal contiene $n+1$ números, que son los coeficientes $\binom{n}{0}$, $\binom{n}{1}$, y así sucesivamente hasta $\binom{n}{n}$. De tal manera que las primeras líneas son: 
$$1 1$$
$$1 2 1$$
$$1 3 3 1$$
$$1 4 6 4 1$$

```python
def binomial(n,k):

    if k == 0:
        return 1

    numerador = 1
    denominador = 1

    for i in range(1,k+1):
        numerador = numerador*(n-i+1)
        denominador = denominador*i

    return numerador//denominador


for n in range(20):

    fila = []

    for k in range(n+1):
        fila.append(binomial(n,k))

    print(fila)
```

#### c) La probabilidad de que para una moneda no sesgada, lanzada $n$ veces, salga águila $k$ veces es:

$$p(k\vert{}n) = \frac{\binom{n}{k}}{2^n}$$

#### Escribe un programa para calcular:
#### 1) La probabilidad total de que una moneda lanzada 100 veces, salga águila exactamente 60 veces y
#### 2) La probabilidad de que salga águila 60 veces o más.

```python
def binomial(n,k):

    if k == 0:
        return 1

    numerador = 1
    denominador = 1

    for i in range(1,k+1):
        numerador = numerador*(n-i+1)
        denominador = denominador*i

    return numerador//denominador


probabilidad_60 = binomial(100,60)/2**100

print("Probabilidad de obtener exactamente 60 águilas:")
print(probabilidad_60)


probabilidad_60_o_mas = 0

for k in range(60,101):
    probabilidad_60_o_mas = probabilidad_60_o_mas + binomial(100,k)/2**100

print("Probabilidad de obtener 60 águilas o más:")
print(probabilidad_60_o_mas)
```

### **4. Números primos**
#### Es posible escribir un programa bastante rápido para calcular números primos siguiendo las siguientes observaciones:
#### a) Un número $n$ es primo, si no tiene factores primos menores que $n$. Por lo tanto, sólo necesitamos comprobar si es divisible por otros números primos.


#### b) Si un número $n$ no es primo y tiene un factor $r$, entonces $n = rs$, donde $s$ también es un factor. Si $r \ge \sqrt{n}$ entonces $n = rs \ge \sqrt{n}s,$ lo que implica que $s \le \sqrt{n}.$ Es decir, cualquier no primo debe tener factores, y por lo tanto también factores primos, menores o iguales a $\sqrt{n}.$ Por lo tanto, para determinar si un número es primo, tenemos que verificar sus factores primos solo hasta $\sqrt{n};$ incluso si no hay ninguno, entonces el número es primo.

#### c) Si encontramos un factor primo menor que $\sqrt{n}$ entonces sabemos que el **número no es primo** y, por lo tanto, no hay necesidad de realizar más comprobaciones; podemos abandonar este número y pasar al siguiente.

#### Escribe un programa que encuentre todos los números primos hasta 10 000 (diez mil). Crea una lista para almacenar los números primos, que comience con solo un número primo (el 2). Luego, para cada número $n$ del 3 al 10 000, verifica si el número es divisible por alguno de los números primos de la lista hasta $\sqrt{n}.$ En cuanto encuentres un factor primo, puedes dejar de comprobar el resto de números (pues ya sabrías que $n$ no es primo). Si no encuentras factores primos iguales o menores a $\sqrt{n},$ entonces $n$ es primo y habría que agregarlo a la lista. Puedes imprimir la lista completa al final del programa, o bien, ir imprimiendo los números individualmente a medida que los vayas encontrando.

```python
from math import sqrt

primos = [2]

for n in range(3,10001):

    es_primo = True

    for p in primos:

        if p > sqrt(n):
            break

        if n%p == 0:
            es_primo = False
            break

    if es_primo == True:
        primos.append(n)

print(primos)
```

### **5. Recursión**

#### a) Anteriormente, en clase encontramos los números de Catalán $C_n$. Si hacemos una pequeña modificación, la definición de $C_n$ se puede reescribir de la siguiente forma:

$$C_n = \begin{cases} 
1 & \text{si } n = 0, \\ 
\frac{4n-2}{n+1} C_{n-1} & \text{si } n > 0. 
\end{cases}$$

#### Escribe una función, usando **recursividad**, que calcule $C_n$ y utilízala para calcular e imprimir $C_{100}$.


```python
def catalan(n):

    if n == 0:
        return 1

    else:
        return ((4*n-2)/(n+1))*catalan(n-1)


print(catalan(100))
```

#### b) Euclides demostró que el máximo común divisor $g(m,n)$ de dos enteros no negativos $m$ y $n$ satisface que:

$$g(m,n) = \begin{cases} 
m & \text{si } n = 0, \\ 
g(n, m \bmod n) & \text{si } n > 0. 
\end{cases}$$
#### Usando esta fórmula escribe una función $g(m,n)$ que emplee **recursividad** para calcular el máximo común divisor de $m$ y $n$. Usa tu función para calcular e imprimir el máximo común divisor de 108 y 192.

```python
def g(m,n):

    if n == 0:
        return m

    else:
        return g(n,m%n)


print(g(108,192))
```
